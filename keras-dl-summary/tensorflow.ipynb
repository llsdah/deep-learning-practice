```python
import numpy as np

x = np.linspace(0, 1, 5)
print(x)
print(type(x), x.shape, x.dtype)
```

```python
import tensorflow as tf

x = tf.linspace(0, 1, 5)
print(x)
print(type(x), x.shape, x.dtype)
```

```python
x = list(range(12))
print(x)
```

```python
x = np.arange(12)
print(x)
x = x.reshape(3, 4)
print(x)
```

```python
x = tf.range(12)
print(x)
# x = x.reshape((3, 4)) # TensorFlow에서는 reshape이 아니라 tf.reshape 함수를 사용해야 합니다.
x = tf.reshape(x, shape=(3, 4))
print(x)
```
